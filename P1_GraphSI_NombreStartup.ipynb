## 1. Problema del cliente

Una empresa comercial utiliza un sistema de planificación de recursos empresariales (ERP) para gestionar sus principales actividades, entre ellas ventas, compras, inventario y contabilidad. El sistema está compuesto por diferentes módulos y pantallas que se encuentran conectados mediante opciones de navegación y son utilizados por empleados con distintas responsabilidades, como vendedores, contadores y gerentes.

La empresa necesita conocer cómo está organizada la navegación de su ERP y determinar si sus usuarios pueden acceder correctamente a las pantallas y funcionalidades que requieren para realizar sus actividades. Actualmente, no cuenta con un análisis que permita identificar las conexiones existentes entre los módulos, conocer las rutas disponibles para llegar a una funcionalidad o establecer si los recorridos realizados representan la alternativa más directa.

Esta necesidad plantea varias interrogantes: ¿Es posible llegar desde una pantalla inicial hasta cualquier funcionalidad requerida? ¿Cuál es la ruta que necesita menos pasos? ¿Qué pantallas pueden alcanzarse desde un módulo determinado? ¿Existen rutas alternativas para acceder a una misma funcionalidad? ¿Cómo afectan los ciclos y las modificaciones de las conexiones a la navegación del sistema?

Para responder a estas preguntas, "Colocar nombre"" propone desarrollar un prototipo analítico en Python que represente la estructura de navegación del ERP mediante un grafo dirigido y no ponderado. Cada vértice representará un módulo o pantalla, mientras que las aristas representarán las conexiones que permiten desplazarse entre ellos.

Sobre esta representación se implementarán los algoritmos BFS y DFS con lógica propia. BFS permitirá encontrar rutas con la menor cantidad de conexiones entre un origen y un destino, mientras que DFS permitirá explorar las pantallas alcanzables desde un punto de partida e identificar aquellas que no pueden visitarse. Adicionalmente, se realizarán pruebas y modificaciones controladas en las conexiones del grafo para evaluar cómo cambia la accesibilidad y la cantidad de pasos necesarios para llegar a determinadas funcionalidades.

### Decisiones de modelado
| Elemento | Decisión | Justificación |
|---|---|---|
| **Vértice** | Una pantalla o destino de navegación del ERP | Es la unidad a la que el usuario llega; da granularidad suficiente para analizar rutas. |
| **Arista** | Una transición directa (un clic) de una pantalla a otra | Cada arista equivale a un paso de navegación. |
| **Tipo de grafo** | **Dirigido** | Poder ir de A a B no implica poder volver de B a A. Así existen pantallas no alcanzables desde ciertos puntos de entrada. |
| **Ponderación** | **No ponderado** | Cada arista es un paso; el criterio de "camino corto" es el número de aristas. |
| **Punto de entrada principal** | `Menú principal` (tras `Login`) | Es el origen natural de la navegación. |
